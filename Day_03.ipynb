```python
import matplotlib.pyplot as plt
```

```python
# Line plot
year = [2019, 2020, 2021, 2022, 2023]
sales = [300, 200, 400, 700, 100]

plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Year vs Sales')

plt.plot(year, sales, color='blue', marker='o')
plt.show()
```

```python
#bar chart
plt.bar(year, sales, color='blue')
plt.show()
```

```python
plt.hist(sales, bins=5, color='blue', edgecolor='black')

plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Sales Distribution')

plt.show()
```

SEABORN

```python
import seaborn as sns
```

```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

```python
penguins.isnull().sum()
```

```python
#distribution plot
sns.displot(data=penguins, x="flipper_length_mm")
plt.show()
```

```python
sns.displot(data=penguins, x="bill_length_mm")
plt.show()
```

```python
penguins['sex'].value_counts().plot(kind='bar')
```

```python
#plotting sex wise species count using seaborn
sns.barplot(data = penguins, x='species', hue='sex', palette='rocket' )
```

```python
sns.countplot(data = penguins, x='species', hue='sex', palette='coolwarm')
```

```python
#categorical plots
sns.catplot(kind = 'box', data = penguins, x='species', y='bill_length_mm')
plt.show()
```

```python
sns.catplot(kind = 'swarm', data = penguins, x='species', y='bill_length_mm', hue='sex')
plt.show()
#stripplot()(with kind ="strip"; the default)
#swarmplot()(with kind="swarm")
#Categorical distribution plots:
#boxplot() (with kind="box")
#violinplot() (with kind="violin")
#boxplot() (with kind="boxen")
```
