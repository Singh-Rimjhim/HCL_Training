```python
#Task : Which  Product Generated the highest revenue ?
#Calulate the revenue for each product
#Sort products by revenue in descending order
#Identify the top three revenue generating products
#Create a bar chart showimg revenue by product
#Display the revenue on the bars
#Create an interactive plotly bar chart with hover information showing Product name and Revenue
```

```python
#Task : Which  Product Generated the highest revenue ?
#Calulate the revenue for each product
product_revenue = (
    sales.groupby('Product')['Total_Amount_INR']
    .sum()
    .sort_values(ascending=False)
)
product_revenue.head(3)
```

```python
# Sort products by revenue in descending order
product_revenue = product_revenue.sort_values(ascending=False)
product_revenue.head()
```

```python
#Identify the top three revenue generating product
top_products = product_revenue.head(3)
top_products
```

```python
#Create a bar chart showing revenue by product

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.bar(product_revenue.index, product_revenue.values, color='skyblue')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.title('Product Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Display the revenue on the bars
#Create an interactive plotly bar chart with hover information showing Product name and Revenue

fig = px.bar(product_revenue, x=product_revenue.index, y=product_revenue.values, labels={'x': 'Product', 'y': 'Revenue'}, title='Product Revenue')
fig.show()
```

```python
#Task 2 : Which Cities Generated the Most Sales ?

#Calulate the total sales for each city
city_sales = sales.groupby('City')['Total_Amount_INR'].sum().sort_values(ascending=False)
city_sales
```

```python
#Rank cities from highest to lowest revenue
city_sales = city_sales.rank(ascending=False)
city_sales
```

```python
# Create a horizontal bar chart showing revenue by city
fig = px.bar(city_sales, x=city_sales.values, y=city_sales.index, orientation='h', labels={'x': 'Revenue', 'y': 'City'}, title='City Revenue')
fig.show()
```

```python
#Bonus question
revenue_by_city = sales.groupby('City', as_index=False)['Total_Amount_INR'].sum()

fig = px.bar(
    revenue_by_city,
    x='City',
    y='Total_Amount_INR',
    title='Total Revenue by City',
    labels={
        'City': 'City',
        'Total_Amount_INR': 'Total Revenue (INR)'
    },
    hover_data={'Total_Amount_INR': ':,.2f'}
)

fig.update_layout(
    xaxis_title='City',
    yaxis_title='Total Revenue (INR)',
    xaxis_tickangle=-45
)

fig.show()
```

```python
#Ques:-4
#Question 4: Which Sales Channel Performs Best?
#Business problem: The company sells through multiple channels and wants to compare their performance
#Tasks:
#Calculate the number of orders for each sales channel
#Calculate total revenue generated by each channel
#Compare order counts and revenue.
#Identify the channel with the highest revenue
#Determine whether the channel with the most orders also generates the highest revenue
#Visualization: Bar chart
#Libraries: Matplotlib and Seaborn
#Bonus: Create a Plotly grouped bar chart comparing order count and revenue. Use separate charts if the different scales make the comparison difficult to interpret
```

```python
# Calculate the number of orders for each sales channel
orders = sales.groupby("Sales_Channel")["Order_ID"].count()

# Display the result
print(orders)
```

```python
# Calculate total revenue generated by each sales channel
revenue = sales.groupby("Sales_Channel")["Total_Amount_INR"].sum()
print(revenue)
```

```python
# Calculate unique order counts and total revenue
summary = sales.groupby("Sales_Channel").agg(
    Order_Count=("Order_ID", "nunique"),
    Total_Revenue=("Total_Amount_INR", "sum")
).reset_index()

print(summary)

# Compare order counts
plt.figure(figsize=(8, 5))
sns.barplot(data=summary, x="Sales_Channel", y="Order_Count")
plt.title("Order Count by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Compare revenue
plt.figure(figsize=(8, 5))
sns.barplot(data=summary, x="Sales_Channel", y="Total_Revenue")
plt.title("Revenue by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Total Revenue (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Identify the channel with the highest revenue
revenue = sales.groupby("Sales_Channel")["Total_Amount_INR"].sum()

highest_revenue_channel = revenue.idxmax()
highest_revenue = revenue.max()

print("Channel with highest revenue:", highest_revenue_channel)
print("Total revenue:", highest_revenue)
```

```python
# Find the channel with the most orders
orders = sales.groupby("Sales_Channel")["Order_ID"].nunique()
most_orders_channel = orders.idxmax()

# Find the channel with the highest revenue
revenue = sales.groupby("Sales_Channel")["Total_Amount_INR"].sum()
highest_revenue_channel = revenue.idxmax()

print("Channel with most orders:", most_orders_channel)
print("Channel with highest revenue:", highest_revenue_channel)

# Compare both channels
if most_orders_channel == highest_revenue_channel:
    print("Yes, the channel with the most orders also generates the highest revenue.")
else:
    print("No, the channel with the most orders does not generate the highest revenue.")
```

```python

# Calculate order count and revenue by sales channel
summary = sales.groupby("Sales_Channel").agg(
    Order_Count=("Order_ID", "nunique"),
    Total_Revenue=("Total_Amount_INR", "sum")
).reset_index()

# Plotly bar chart for order count
fig1 = px.bar(
    summary,
    x="Sales_Channel",
    y="Order_Count",
    title="Order Count by Sales Channel",
    text_auto=True,
    color="Sales_Channel"
)

fig1.show()

# Plotly bar chart for revenue
fig2 = px.bar(
    summary,
    x="Sales_Channel",
    y="Total_Revenue",
    title="Total Revenue by Sales Channel",
    text_auto=True,
    color="Sales_Channel"
)

fig2.show()
```
