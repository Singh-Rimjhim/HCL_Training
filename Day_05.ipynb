```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as plx
```

```python
df = pd.read_csv('hr_employee_dataset.csv')
print(df.head)
```

```python
#Display the numbers of row and column
print("Numbers of rows: ", df.shape[0])
print("Numbers of column: ", df.shape[1])
print("Dataset Shape: ", df.shape)
```

```python
#Identify missing values in each column
print(df.isnull().sum())
```

```python
#Find and remove duplicate record
print("Duplicate rows: ", df.duplicated().sum())
print("Duplicate records: ")
print(df[df.duplicated()])
df=df.drop_duplicates()
print("Shape after removing duplicates: ", df.shape)
```

```python
#Identify duplicate employee ids
print(df[df.duplicated(subset=['employee_id'], keep=False)])
```
